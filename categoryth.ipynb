# Category Theory

A **point** of a set $X$ is a map $1 \to X$.


**Endofunctor**: A functor that maps a category to itself.

A category with one Element is called a **monoid**.


A monad is used to represent a sequence of steps of computation.

A monad is a way to abstract a control-flow effect, so that users write a series of pure functions sequences between the effect points of their program. So instead of writing a section of code which contains effect points embedded in it, monads allow users to write a series of pure functions strung together across those effect points.
